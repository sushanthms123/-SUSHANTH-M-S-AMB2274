```python
rows=int(input("enter the number of rows:"))
for i in range(rows+1):
  print("*" * (i))
```

```python
height=int(input("enter the height:"))
for i in range(2,height):
  print(" " * (height - i - 1), end="")
  print("*" * (2 * i + 1))
```
