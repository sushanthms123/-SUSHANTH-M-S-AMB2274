```python
marks = input("Enter marks separated by commas: ")
marks = [int(i) for i in marks.split(',')]
maximum_value = __builtins__.max(marks)
minimum_mark=min(marks)
average=sum(marks)/len(marks)
print(f"maximum mark:{maximum_value}, minimum mark:{minimum_mark}, average mark:{average}")
```
